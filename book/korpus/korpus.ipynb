## 📝 Untertitel und Metadaten herunterladen

Im Folgenden finden Sie eine Zusammenfassung der Erstellung des gesamten Korpus als sogenannter Pseudocode. Pseudocode bedeutet in diesem Fall die chronologische Aufzählung der für die Erstellung nötigen Schritte. Diese Beschreibung verzichtet dabei soweit wie möglich auf technisches Vokabular der Programmiersprache Python, um die Zugänglichkeit zu erhöhen. In Abschnitt {ref}`🚀 Korpus erstellen<scraping>` finden Sie ein herunterladbares Juypter Notebook, um das in {cite:t}`Scholl.imDruck` qualitativ analysierte Korpus zu reproduzieren.

__Anmerkung__: Um eine Belastung des Servers zu vermeiden, empfiehlt es sich jeglichen Code zur Datenaquise am Abfragelimit der [Tagesschau API (V. 2.0.1)](https://tagesschau.api.bund.dev/) (d. h. max. 60 Abfragen pro Minute) auszurichten. Bei dem hier ausgeführten Code, läuft der entsprechende Code bis zu 20 Stunden. Es empfiehlt sich also, den übergeordneten for-Loop auf mehrere Teile aufzuteilen und den Code über Nacht laufen zu lassen.

```{code}
Pseudocode
----------
Task: Links und Metadaten für das zu erstellende Korpus sammeln (scrapen)

- Links für die gewünschten Videos (Landingpage) aus der Mediathek sammeln
    * URL für dynamisch nachgeladene Inhalte je nach Untersuchungszeitraum definieren
    https://api.ardmediathek.de/page-gateway/widgets/ard/asset/Y3JpZDovL2Rhc2Vyc3RlLmRlL3RhZ2Vzc2NoYXU?pageNumber={Seitenzahl}&pageSize={Seitenumfang}
    * Dataframe mit Spalten für Links, Daten und gewünschten Metadaten erstellen
    * Ende des Loops für die Abfrage dynamisch geladener Inhalte festlegen
    * Sleeptimer entsprechend dem Abfragelimits festlegen

- Links zu den Untertiteldateien (VTT) + Metadaten sammeln.
    * Sleeptimer entsprechend dem Abfragelimits festlegen.
    * Loop durch die oben gesammelten Links zu Videos
        + HTML-Quellcode mittels HTTP-Bibliothek (z. B. requests) abfragen
        + Für jedes Datum im Dataframe Links zur Untertiteldatei, Themenübersicht, Titel aus JSON kopieren
            --> Achtung: nicht alle Videos sind Untertitelt; fehlerhafter Titel oder falsches Datum auf der Landingpage können zu Fehlern führen
            --> Der Titel ist wichtig, um die richtigen Inhalte herauszufiltern (z. B. nur 20-Uhr-tagesschau)
        + Bei multimodalen Studien kann der Link zur Landingpage genutzt werden, um später Videoausschnitte mit den Untertiteln zu alignieren
```

Ein ausführlicherer Pseudocode der Erstellung eines Korpus, das um die Anschlusskommunikation zu den Sendungen der _tagesschau_ in YouTube-Kommentaren erweitert wird, wird als Begleitmaterial zu {cite:t}`Scholl.imDruckb` veröffentlicht.
