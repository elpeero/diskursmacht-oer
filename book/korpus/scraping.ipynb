Da es sich hier nur um 25 VTT-Dateien handelt, wird das selbstauferlegt Abgfragelimit von 60 Abfragen pro Stunde [(in Anlehnung an die Limit der tagesschau API v2.0.1)](https://tagesschau.api.bund.dev/) bei ein- bis zweimaliger Ausführung des Codes eingehalten.
`````python
# Load libraries
import time
import requests
from datetime import datetime

# Get the subtitle texts
# Initialize log
start_time = datetime.now().strftime('%d-%m-%y_%H-%M')
log_list = [f'Logging started: {str(start_time)}\n\n',]

sub_text_list = []
i = 0

# Get the subtitles
for date in sub_df['Datum']:
    i = i+1
    try:
        sub_url = sub_df.loc[sub_df['Datum'] == date, 'Sub_Link'].values[0]
        if sub_url == 'NA':
            sub_text = 'NA'
            state = 'No subs available'
        else:
            # Request the VTT-text from server
            resp = requests.get(sub_url)
            resp.raise_for_status()
            resp.encoding = 'utf-8'
            sub_text = resp.text
            state = 'successful'

        sub_text_list.append(sub_text)

        if state == 'successful':
            time.sleep(10)

    except Exception as error:
        sub_text_list.append('error')
        state = 'error'
        print(f'Errortype: {type(error)}, text: {error}')
        log_list.append(f'Errortype: {type(error)}, text: {error}')
        continue

    progress = (i*100)/len(sub_df['Datum'])
    print(f'Progress: {progress:.0f}%\t{state}\t{date}')
    log_list.append(f'Progress: {progress:.0f}%\t{state}\t{date}')

# Add the subtitle texts to sub_df
sub_df['Sub_Text'] = sub_text_list

# Log completion
end_time = datetime.now().strftime('%d-%m-%y_%H-%M')
print(f'\nLogging completed: {end_time}')
log_list.append(f'\nLogging completed: {end_time}')

# Write log file
with open(f'scrape_subtitles_log_{start_time}.txt', 'w', encoding='utf-8') as file_out:
    file_out.write("\n".join(log_list))
`````

`````python
import re

def conv_time(time_str):
    """
    Wandelt einen Zeit-String HH:MM:SS[.fff] in Sekunden (float) um.
    Z.B. "00:00:02.960" -> 2.96
    """
    hours, minutes, seconds = time_str.split(':')
    return int(hours) * 3600 + int(minutes) * 60 + float(seconds)

def conv_url(url_in, st, et):
    """
    Erstellt eine URL die den individuellen Frame als Videoausschnitt anzeigt. Um eine wohlgeformte XML erstellen zu können,
    muss das & zwischen startTime und endTime als hmtl-entität eingefügt werden. 
    Dadurch endet das Video aber nicht, sondern läuft einfach weiter. Die Startzeit funktioniert trotzdem.
    -> Um nur den jew. Frame anzuzeigen, die hmtl-entität &amp; wieder durch ein &-Zeichen ersetzen
    """
    video_id = url_in.rstrip('/').split('/')[-1]
    fmt = f'{{st:.2f}}'  
    url_out = (
        f'https://www.ardmediathek.de/embed/{video_id}?startTime={fmt.format(st=st)}&endTime={fmt.format(st=et)}'
    )
    return url_out

def find_section(ts_str, parsed_sections):
    """"
    Ordnet eine Zeit einem Beitragstitel zu.
    """
    ts_time = datetime.strptime(ts_str, "%H:%M:%S.%f").time()
    for i in range(len(parsed_sections) - 1, -1, -1):
        if parsed_sections[i][0] <= ts_time:
            sections = parsed_sections[i][1].rstrip(',')
            return sections
        
    return None

# Create lists to temorarily store the parsed information
date_ls = []
v_url_ls = []
sec_no_ls = []
sec_title_ls = []
frame_no_ls = []
frame_text_ls = []
time_ls = []

# Iterate through dataframe with VTT source material
for _, row in sub_df.iterrows():

    date = row['Datum']
    v_url = row['Video_Link']
    sections = row['Sections']
    parsed_sections = [(datetime.strptime(t, "%M:%S").time(), title) for t, title in sections]
    
    # Prepare parsing of sections
    prev_sec_title = None
    # Set section numbering to default 0 for every broadcast
    section_counter = 0

    # Parse VTT-text
    content = row['Sub_Text']
    content = re.sub(r'([\[\-\]])', r' \1 ', content)
    content = re.sub(r'([0-9]{2}:[0-9]{2}:[0-9]{2}\.[0-9]{3}).+?([0-9]{2}:[0-9]{2}:[0-9]{2}\.[0-9]{3}).+', r'\1-\2', content)
    content = re.sub(r'\n\n\n', r'\n\n', content)
    blocks = content.split("\n\n")

    # Split content into frame-blocks
    for block in blocks:
        if re.search(r'(^\d+)\n([^\n]+)\n<c\.text([^>]+)(.+)', block, re.DOTALL):
            in_block = re.search(r'(^\d+)\n([^\n]+)\n(<c\.text[^>]+.+)', block, re.DOTALL)

            # Create linkt to video frame
            time_stamp = in_block.group(2)
            beg, end = time_stamp.split("-")

            sec_title = find_section(beg, parsed_sections)
            
            # Parse sections
            if sec_title != "NA" and sec_title != prev_sec_title:    
                # Number each section
                section_counter = section_counter + 1
                prev_sec_title = sec_title     
            conv_beg = conv_time(beg)
            conv_end = conv_time(end)
            f_url = conv_url(v_url, conv_beg, conv_end)      # link to video frame
            f_no = in_block.group(1)
            t_lines = in_block.group(3).split('\n')
            lines_ls = []
            for line in t_lines:
                f_text = re.search(r'>([^<]+)', line).group(1)
                lines_ls.append(f_text)
            f_text = (' ').join(lines_ls)
            date_ls.append(date)
            v_url_ls.append(f_url)
            sec_no_ls.append(section_counter)
            sec_title_ls.append(sec_title)
            frame_no_ls.append(f_no)
            frame_text_ls.append(f_text)
            time_ls.append(time_stamp)

# Create dataframe with corpus data
corpus_data = {
    'Datum' : date_ls,
    'Beitragsnummer' : sec_no_ls,
    'Beitragstitel' : sec_title_ls,
    'Framenummer' : frame_no_ls,
    'Frame-URL' : v_url_ls,
    'Zeitstempel' : time_ls,
    'Text' : frame_text_ls
}

corpus_df = pd.DataFrame(corpus_data)

# Export dataframe to the current working directory i. e. the directory where the jupyter notebook lives
corpus_df.to_csv('Korpus.txt', sep='\t', encoding='utf-8')
`````
